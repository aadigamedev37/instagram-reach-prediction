```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import shutil
import os

# Ensure figures/final exists
os.makedirs("../reports/figures/final", exist_ok=True)

# We will copy the most important charts generated in EDA and Model Analysis into the final folder.
charts_to_copy = [
    "reach_by_day.png",
    "reach_by_post_type.png",
    "feature_importance.png",
    "actual_vs_predicted.png"
]

for chart in charts_to_copy:
    src = f"../reports/figures/{chart}"
    dst = f"../reports/figures/final/{chart}"
    if os.path.exists(src):
        shutil.copy(src, dst)
        print(f"Copied {chart} to final/")
    else:
        print(f"File {chart} not found in figures/")
```
