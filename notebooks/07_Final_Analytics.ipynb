# Notebook 07: Final Analytics

## 1. Dataset Summary
We analyzed a dataset of 50,000 Instagram posts.

## 2. EDA Findings
We discovered that Post Type and the Day of the Week have meaningful associations with Reach.

## 3. Modeling
We trained multiple models to predict Expected Post Reach using features available prior to publishing (e.g., Post Type, Day, Hashtags). Random Forest / Gradient Boosting emerged as the strongest performers.

## 4. Recommendation System
The system evaluates the 7 days of the week for a proposed post to identify which day historically yields the highest predicted reach.

## 5. Conclusion
The project successfully demonstrates the use of machine learning to provide data-driven posting recommendations.
